```python
import pandas as pd
import pickle 
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python
path = ["results/datasets/Arc_lens_image_list.pickle", "results/datasets/Non_lens_image_list.pickle", "results/datasets/Ring_image_list.pickle", "results/datasets/Multi_lens_image_list.pickle"]

with open('results/datasets/Arc_lens_image_list.pickle', 'rb') as f:
    images = pickle.load(f)

images = np.arcsinh(images)

for k in images:
    for layer in range(len(k)):
        k[layer] = k[layer] - k[layer].min()
        k[layer] = k[layer] / k[layer].max()
        
    pic = np.moveaxis(k, 0, -1)            
    plt.imshow(pic, origin='lower', interpolation='bicubic')
    plt.xlim(30, 70)    # show only columns 30–70
    plt.ylim(30, 70)    # show only rows 30–70
    plt.show()
```
