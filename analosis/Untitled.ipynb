```python
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

st.title("Dice roller")                                   # big heading at the top

n_rolls = st.number_input("How many rolls?", min_value=1, max_value=10000, value=100)
seed = st.number_input("Seed", value=42)

if st.button("Roll"):                                     # everything below only runs when clicked
    with st.spinner("Rolling..."):                        # shows a spinner while this block runs
        rng = np.random.default_rng(int(seed))
        rolls = rng.integers(1, 7, size=int(n_rolls))

    st.success(f"Rolled {len(rolls)} dice")               # green message box
    st.metric("Average", f"{rolls.mean():.2f}")           # a big number display

    fig, ax = plt.subplots()
    ax.hist(rolls, bins=np.arange(0.5, 7.5, 1))
    ax.set_title("Results")
    st.pyplot(fig)                                        # shows a matplotlib figure

    counts = pd.Series(rolls).value_counts().sort_index()
    st.dataframe(counts)                                  # shows a table
```
